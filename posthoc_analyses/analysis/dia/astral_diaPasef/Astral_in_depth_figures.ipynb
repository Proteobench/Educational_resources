# Astral in-depth figures

A 3-row x 2-column figure built entirely with Plotly:

- **a** (top-left): main Astral performance scatter (all points) — same as composite-figure panel a.
- **b** (top-right): same scatter, restricted to a subset of points (hashes filled in below).
- **c** (middle-left): same scatter, restricted to a second subset of points (hashes filled in below).
- **d** (middle-right): CV violin plot for one specified run.
- **e** (bottom-left): MA plot for that same run.
- **f** (bottom-right): log2 fold-change distribution for that same run.

Panels d/e/f come directly from `LFQHYEPlotGenerator.generate_in_depth_plots()` — the same
in-depth plots shown in the webinterface's "View Single Result" tab.

This notebook re-extracts the Astral submissions from scratch (with pickle caching), so an
internet connection is required unless a cached `.pkl` is already present.

## Helper functions

```python
def extract_all_from_hashes(
    hashes, data_dir="extracted_files", module=DIAQuantIonModuleAstral(token="")
):
    """Extract all results from hashes."""
    all_datapoints = pd.DataFrame()
    results_dict = {}

    for point_hash, tool in hashes.items():
        print(f"Processing {point_hash} - {tool}")
        if not tool in results_dict:
            results_dict[tool] = {}

        raw_data_path = f"{data_dir}/{point_hash}/input_file"  # Extension can vary
        param_path = f"{data_dir}/{point_hash}/param_0."  # Extension can vary
        if tool == "AlphaDIA":
            secondary_raw_data_path = (
                f"{data_dir}/{point_hash}/input_file_secondary"  # Extension can vary
            )

        # Identify the correct raw data file extension. There can be only one raw data file per point_hash.
        raw_data_files = list(
            pathlib.Path(raw_data_path).parent.glob(
                f"{pathlib.Path(raw_data_path).name}.*"
            )
        )
        print(f"Raw data files found: {raw_data_files}")
        # Identify the correct parameter file extension. There can be only one parameter file per point_hash.
        param_files = list(
            pathlib.Path(param_path).parent.glob(f"{pathlib.Path(param_path).name}.*")
        )
        print(f"Parameter files found: {param_files}")

        if tool == "AlphaDIA":
            secondary_raw_data = list(
                pathlib.Path(secondary_raw_data_path).parent.glob(
                    f"{pathlib.Path(secondary_raw_data_path).name}.*"
                )
            )
            print(f"Secondary raw data files found: {secondary_raw_data}")
            if not secondary_raw_data:
                print(
                    f"Skipping {point_hash} - {tool}: Reverting to primary raw data only - No secondary raw data file found."
                )
                secondary_raw_data_path = None
            else:
                secondary_raw_data_path = secondary_raw_data[0]

        if not raw_data_files or not param_files:
            print(
                f"Skipping {point_hash} - {tool}: No raw data or parameter files found."
            )
            continue
        raw_data_path = raw_data_files[0]
        param_path = param_files[0]

        if "FragPipe" in tool or "AlphaPept" in tool:
            param_path = open(param_path, "rb")

        extract_params_func = globals()[
            f"extract_{tool.replace('-', '').lower().replace(' (diann quant)', '')}_params"
        ]
        param_data = extract_params_func(param_path).__dict__
        results_performance, all_datapoints, _ = module.benchmarking(
            raw_data_path,
            tool,
            user_input=param_data,
            all_datapoints=all_datapoints,
            input_file_secondary=(
                secondary_raw_data_path if tool == "AlphaDIA" else None
            ),
        )

        if "MaxQuant" in tool:
            print("adding carbamidos to maxquant")
            results_performance["precursor ion"] = results_performance[
                "precursor ion"
            ].apply(lambda x: x.replace("C", "C[Carbamidomethyl]"))

        results_dict[tool][point_hash] = results_performance
    # Set all datapoints to old
    all_datapoints["old_new"] = "old"
    return results_dict, all_datapoints
```

```python
def prepare_datapoints(all_datapoints, min_quant=3):
    """Filter and calculate metrics for data points at a given min_quant (k) threshold.

    Column names match exactly what `LFQHYEPlotGenerator.plot_main_metric` reads
    (`nr_feature`, `{metric}_abs_epsilon_{global,eq_species}`).
    """
    for metric in ("median", "mean"):
        for mode in ("global", "eq_species"):
            all_datapoints[f"{metric}_abs_epsilon_{mode}"] = [
                filter_df_numquant_epsilon(v, min_quant=min_quant, metric=metric, mode=mode)
                for v in all_datapoints["results"]
            ]
    all_datapoints["nr_feature"] = [
        filter_df_numquant_nr_feature(v, min_quant=min_quant) for v in all_datapoints["results"]
    ]
    return all_datapoints


def plot_performance_metrics(
    all_datapoints,
    min_quant=3,
    metric="Median",
    subset=None,
    x_range=None,
    y_range=None,
    subset_col="id",
):
    """Plot performance metrics for data points."""
    if subset:
        all_datapoints = all_datapoints[all_datapoints[subset_col].isin(subset)]
    all_datapoints = prepare_datapoints(all_datapoints, min_quant=min_quant)
    plot = LFQHYEPlotGenerator().plot_main_metric(
        all_datapoints,
        metric=metric,
        mode="Species-weighted",
        label="None",
        legend_name_map={"AlphaPept": "AlphaPept (legacy tool)", **_DISPLAY_NAMES},
    )
    plot.update_layout(template="plotly_white")
    x_dict = dict(
        title=f"{metric}QuantError({min_quant})",
        title_font=dict(size=16),
        tickfont=dict(size=15),
    )
    if x_range:
        x_dict["range"] = x_range
    y_dict = dict(
        title=f"Sensitivity({min_quant})",
        title_font=dict(size=16),
        tickfont=dict(size=15),
    )
    if y_range:
        y_dict["range"] = y_range
    plot.update_layout(xaxis=x_dict, yaxis=y_dict)
    return plot
```

## Build the main figure

```python
# Species-expected-ratio lookup (needed for the MA / log2FC panels). Only depends on the
# module's shared settings, so any tool present in the Astral module's parser mapping works.
astral_module = DIAQuantIonModuleAstral(token="")
parse_settings = ParseSettingsBuilder(
    parse_settings_dir=astral_module.parse_settings_dir,
    module_id=astral_module.module_id,
).build_parser("DIA-NN")
```

```python
# ---- Panels d/e/f: CV violin, MA plot, log2FC distribution for the selected run ----
_species_expected_ratio = parse_settings.species_expected_ratio()
_species_colors = {sp: d["color"] for sp, d in _species_expected_ratio.items()}


def _fix_expected_annotations(fig, species_colors):
    """Rename 'Expected SPECIES' vline annotations to just 'SPECIES', colored to match."""
    for ann in fig.layout.annotations:
        if isinstance(ann.text, str) and ann.text.startswith("Expected "):
            species = ann.text.replace("Expected ", "", 1)
            ann.update(text=species, font=dict(color=species_colors.get(species, "black")))


_indepth_plots = LFQHYEPlotGenerator().generate_in_depth_plots(
    run_df.copy(), parse_settings=parse_settings
)
_fig_d = _indepth_plots["cv"]
_fig_f = _indepth_plots["logfc"]

# Panel e is built separately (not via generate_in_depth_plots) so the outlier
# removal below only affects this panel, not the CV/log2FC panels.
_lo, _hi = run_df["log2_A_vs_B"].quantile([MA_OUTLIER_FRACTION / 2, 1 - MA_OUTLIER_FRACTION / 2])
_ma_df = run_df[run_df["log2_A_vs_B"].between(_lo, _hi)].copy()
_fig_e = LFQHYEPlotGenerator()._plot_ma_plot(_ma_df, _species_expected_ratio)
_fig_e.update_layout(
    xaxis=dict(title="log2_FC(A:B)"),
    yaxis=dict(title="Intensity(log)"),
)
_fig_f.update_layout(
    xaxis=dict(title="log2_FC(A:B)"),
    yaxis=dict(title="Density"),
)

_fix_expected_annotations(_fig_e, _species_colors)
_fix_expected_annotations(_fig_f, _species_colors)

for _fig in (_fig_d, _fig_e, _fig_f):
    _fig.update_layout(
        font=dict(size=20),
        xaxis=dict(
            tickfont=dict(size=19), title_font=dict(size=21),
            showline=True, linecolor="black", linewidth=1,
        ),
        yaxis=dict(
            tickfont=dict(size=19), title_font=dict(size=21),
            showline=True, linecolor="black", linewidth=1,
        ),
    )

# Force horizontal gridlines (not guaranteed by the library's default styling for these panels)
_fig_d.update_yaxes(showgrid=True, gridcolor="lightgray", gridwidth=1)
_fig_e.update_yaxes(showgrid=True, gridcolor="lightgray", gridwidth=1)
```

```python
# ---- Assemble the 3x2 grid ----
combined_fig = make_subplots(
    rows=3,
    cols=2,
    horizontal_spacing=0.10,
    vertical_spacing=0.08,
)

_panel_positions = {
    "a": (_fig_a, 1, 1),
    "b": (_fig_b, 1, 2),
    "c": (_fig_c, 2, 1),
    "d": (_fig_d, 2, 2),
    "e": (_fig_e, 3, 1),
    "f": (_fig_f, 3, 2),
}
_subplot_n = {"a": 1, "b": 2, "c": 3, "d": 4, "e": 5, "f": 6}


def _axis_ref(n, letter):
    return letter if n == 1 else f"{letter}{n}"


def _apply_axis(src, dst_key, fig):
    j = src.to_plotly_json()
    for k in ("domain", "anchor", "matches"):
        j.pop(k, None)
    if j:
        fig.update_layout(**{dst_key: j})


def _remap_ref(ref, old_n, new_n):
    if not ref:
        return ref
    old_x, old_y = _axis_ref(old_n, "x"), _axis_ref(old_n, "y")
    new_x, new_y = _axis_ref(new_n, "x"), _axis_ref(new_n, "y")
    return {
        old_x: new_x,
        f"{old_x} domain": f"{new_x} domain",
        old_y: new_y,
        f"{old_y} domain": f"{new_y} domain",
    }.get(ref, ref)


for _label, (_fig, _row, _col) in _panel_positions.items():
    _n = _subplot_n[_label]
    for _tr in _fig.data:
        _tr.update(showlegend=False)  # native legends replaced by the manual swatch legends below
        combined_fig.add_trace(_tr, row=_row, col=_col)

    _apply_axis(_fig.layout.xaxis, _axis_ref(_n, "xaxis"), combined_fig)
    _apply_axis(_fig.layout.yaxis, _axis_ref(_n, "yaxis"), combined_fig)

    for _sh in _fig.layout.shapes or []:
        _j = {k: v for k, v in _sh.to_plotly_json().items() if v is not None}
        _j["xref"] = _remap_ref(_j.get("xref"), 1, _n)
        _j["yref"] = _remap_ref(_j.get("yref"), 1, _n)
        combined_fig.add_shape(**_j)

    for _an in _fig.layout.annotations or []:
        _j = {k: v for k, v in _an.to_plotly_json().items() if v is not None}
        if _j.get("xref") == "paper" and _j.get("yref") == "paper":
            continue  # skip figure-level watermark annotations from plot_main_metric
        _j["xref"] = _remap_ref(_j.get("xref"), 1, _n)
        _j["yref"] = _remap_ref(_j.get("yref"), 1, _n)
        combined_fig.add_annotation(**_j)

    # Panel label, top-left of each panel
    combined_fig.add_annotation(
        text=f"<b>{_label}</b>",
        xref=f"{_axis_ref(_n, 'x')} domain",
        yref=f"{_axis_ref(_n, 'y')} domain",
        x=0,
        y=1,
        xanchor="right",
        yanchor="bottom",
        showarrow=False,
        font=dict(size=25, color="black"),
    )
```

```python
# ---- Manual swatch legends (matching the composite figure's style), stacked at the bottom ----
def _add_swatch_legend(
    fig,
    colors,
    y,
    x_start=0.0,
    x_end=1.0,
    font_size=19,
    display_fn=lambda name: name,
    title=None,
    title_font_size=None,
    title_width=0.09,
):
    if title:
        fig.add_annotation(
            xref="paper",
            yref="paper",
            x=x_start,
            y=y,
            text=f"<b>{title}:</b>",
            showarrow=False,
            font=dict(size=title_font_size or font_size),
            xanchor="left",
            yanchor="middle",
        )
        x_start = x_start + title_width

    n = len(colors)
    span = x_end - x_start
    swatch_w = 14 / 1400
    swatch_h = 14 / 1400
    for i, (name, col) in enumerate(colors.items()):
        xc = x_start + span * (i + 0.5) / n
        fig.add_shape(
            type="rect",
            xref="paper",
            yref="paper",
            x0=xc - swatch_w * 2.8,
            x1=xc - swatch_w * 1.4,
            y0=y - swatch_h / 2,
            y1=y + swatch_h / 2,
            fillcolor=col,
            line_width=0,
            layer="above",
        )
        fig.add_annotation(
            xref="paper",
            yref="paper",
            x=xc - swatch_w * 1.0,
            y=y,
            text=display_fn(name),
            showarrow=False,
            font=dict(size=font_size),
            xanchor="left",
            yanchor="middle",
        )


_tool_colors = {
    name: col
    for name, col in _SOFTWARE_COLORS.items()
    if name in all_datapoints["software_name"].unique()
}

# Species row above the Workflow row, both spanning the full figure width.
_add_swatch_legend(
    combined_fig, _species_colors, y=-0.09, font_size=21, title="Species", title_font_size=21
)
_add_swatch_legend(
    combined_fig,
    _tool_colors,
    y=-0.15,
    font_size=21,
    display_fn=_display,
    title="Workflow",
    title_font_size=21,
)
```

```python
# ---- Global layout: native legends fully replaced by the swatch legends above, save ----
combined_fig.update_layout(
    **_BASE_LAYOUT,
    width=1400,
    height=1500,
    showlegend=False,
    margin=dict(l=80, r=40, t=50, b=220),
)

_out = f"{FIGURES_DIR}astral_in_depth_figure"
try:
    combined_fig.write_image(_out + ".svg")
    combined_fig.write_image(_out + ".png", scale=3)
    print(f"Saved: {_out}.svg / .png")
except Exception as e:
    print(f"Save failed: {e}")

combined_fig.show()
```
